In this project, I performed simple exploratory data analysis on a sample **Titanic dataset**.


```python
# EDA Practice Project: Titanic Dataset

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load Titanic sample data
data = {
    "PassengerId": [1, 2, 3, 4, 5],
    "Survived": [0, 1, 1, 1, 0],
    "Pclass": [3, 1, 3, 1, 3],
    "Sex": ["male", "female", "female", "female", "male"],
    "Age": [22, 38, 26, 35, 35]
}
```

```python
# Convert to dataframe
titanic_df = pd.DataFrame(data)
```

```python
# Display basic info
print(titanic_df.info())
```

```python
# Survivors by Class
survivors_by_class = titanic_df.groupby("Pclass")["Survived"].mean()
print(survivors_by_class)
```

```python
# Visualize survival rate by class
survivors_by_class.plot(kind="bar", title="Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.show()
```
